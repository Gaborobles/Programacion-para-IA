**Ejercicio 1**
Se utilizaran funciones y operadores aritmenticos para evaluar un modelo de clasificación.

```python
def evaluar_modelo():
  valor = input("Dame el valor de la exactitud del modelo (0.0-1.0): ")
  exactitud = float(valor)
  if 1 >= exactitud >= 0.9:
    print("La exactitud del modelo es Excelente 👍")
  elif 0.9 > exactitud >= 0.75:
    print("La exactitud del modelo es Aceptable 👍")
  elif 0.75 > exactitud >= 0.5:
    print("La exactitud del modelo requiere ajuste")
  elif exactitud < 0.5:
    print("La exactitud del modelo no es apto")
```

```python
evaluar_modelo()
```

**Ejercicio 2**
Se utilizaran tuplas, listas y ciclos for para almacenar valores de tasas de aprendizaje.

```python
CONFIGURACION_BASE=(512, 10)
```

```python
def procesar_tasas_aprendizaje(tasas):
  tasas_aprendizaje=[]
  for tasa in tasas:
    tasas_aprendizaje.append(tasa*2.0)
  return tasas_aprendizaje, CONFIGURACION_BASE
```

```python
procesar_tasas_aprendizaje([0.1, 0.01, 0.001, 0.0001])
```

**Ejer**
Se simularan las epocas por las que pasan los algoritmos de aprendizajes.

```python
def simular_entrenamiento():
  cantidad_epocas = int(input("Número maximo de epocas a simular: "))
  i = 1
  while i <= cantidad_epocas:
    perdida = 1 / i
    print("Época", i, "- Pérdida:", perdida)
    if perdida < 0.05:
      print("El modelo ha convergido anticipadamente.")
      break
    i += 1
```

```python
simular_entrenamiento()
```
