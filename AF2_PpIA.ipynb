# Ejercicio 2.2 - Procesamiento de Señales con NumPy

 Se realizan las matrices numericas y se hace cada arreglo pedido, se calculan datos atipicos según su desviación

```python
matriz=(df[['bill_length_mm', 'bill_depth_mm']]).to_numpy()
print(matriz)
proporcion= matriz[:, 0]/matriz[:, 1]

mn_general= np.mean(matriz[:, 0])
std_general= np.std(matriz[:, 0])
outliers= matriz[np.abs(matriz[:, 0]-mn_general)> 2*std_general]
```

```python
print('El radio de largo/ alto del pico de cada pinguino es : ', proporcion)
print()
print('Los outliers son: ', outliers)
```

```python
array_pares=matriz[::2]
print(array_pares)
```

# Ejercicio 2.3.- Visualización Científica con Matplotlib

 Se realizan las visualizaciones requeridas para identificar de una manera más clara diferencias entre tipos de especies y diferencias de características.

```python
plt.figure(figsize=(8, 6))
species_list = df['species'].unique()
colors = {'Adelie': 'red', 'Chinstrap': 'green', 'Gentoo': 'blue'}

for sp in species_list:
    subset = df[df['species'] == sp]
    plt.scatter(subset['flipper_length_mm'], subset['body_mass_g'], color=colors.get(sp, 'black'), label=sp, alpha=0.7)

plt.title("Relación Flipper vs Masa Corporal")
plt.xlabel("flipper_length_mm")
plt.ylabel("body_mass_g")
plt.legend()
plt.grid(True)

idx_max = df['body_mass_g'].idxmax()
peso_max = df.loc[idx_max, 'body_mass_g']
flipper_max = df.loc[idx_max, 'flipper_length_mm']
plt.annotate('Más pesado:', xy=(flipper_max, peso_max), xytext=(flipper_max - 20, peso_max - 500), arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=8), fontsize=10, fontweight='bold')

plt.show()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
for sp in species_list:
    subset = df[df['species'] == sp]
    ax1.scatter(subset['flipper_length_mm'], subset['body_mass_g'],
                color=colors.get(sp, 'black'), label=sp, alpha=0.7)
ax1.set_title("Flipper vs Masa Corporal por Especie")
ax1.set_xlabel("flipper_length_mm")
ax1.set_ylabel("body_mass_g")
ax1.legend()
ax1.grid(True)

ax2.scatter(df['bill_length_mm'], df['flipper_length_mm'], color='purple', alpha=0.6)
ax2.set_title("Largo del Pico vs Tamaño de Aletas")
ax2.set_xlabel("bill_length_mm")
ax2.set_ylabel("flipper_length_mm")
ax2.grid(True)

plt.tight_layout()
plt.show()
```
